```python
import json
from pathlib import Path

train_path = Path("IR/train.json")

with train_path.open("r", encoding="utf-8") as f:
    data = json.load(f)

keys = [
    key
    for key, record in data.items()
    if isinstance(record, dict)
    and isinstance(record.get("answer"), list)
    and len(record["answer"]) > 1
]

print("Total:", len(keys))
keys
```

```python
multi_answer_items = {
    key: record["answer"]
    for key, record in data.items()
    if isinstance(record, dict)
    and isinstance(record.get("answer"), list)
    and len(record["answer"]) > 1
}

print("Total:", len(multi_answer_items))
multi_answer_items
```

```python
import json
from pathlib import Path
from collections import Counter

train_path = Path("IR/train.json")

with train_path.open("r", encoding="utf-8") as f:
    data = json.load(f)

answer_lengths = {
    key: len(record.get("answer", []))
    for key, record in data.items()
}

distribution = Counter(answer_lengths.values())

print("Phân phối số lượng docs trong answer:")
for n_docs, count in sorted(distribution.items()):
    print(f"{n_docs} docs: {count} questions")

distribution
```

```python
import matplotlib.pyplot as plt

xs = sorted(distribution.keys())
ys = [distribution[x] for x in xs]

plt.figure(figsize=(8, 4))
plt.bar(xs, ys)
plt.xlabel("Số lượng docs trong answer")
plt.ylabel("Số lượng questions")
plt.title("Phân phối số lượng docs trong answer - IR/train.json")
plt.xticks(xs)
plt.show()
```
